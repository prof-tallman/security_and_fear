# Caesar Shift Cipher

Encrypting and decrypting with the Caesar Shift Cipher is simple.

````python
shift_key1 = 5
alphabet1 = caesar.shift_alphabet(shift_key1)

print(f'plaintext:  {alphabet1[0]}')     # plaintext (top) is at position 0
print(f'CIPHERTEXT: {alphabet1[1]}\n\n') # ciphertext (bottom) is at position 1
````

````python
plaintext1 = 'This is my test message'
ciphertext1 = caesar.encrypt(plaintext1, shift_key1)

print(f'Encryption: "{plaintext1}" -> "{ciphertext1}"\n\n')
````

````python
ciphertext2 = 'YMNX NX RD YJXY RJXXFLJ'

try2 = caesar.decrypt(ciphertext2, 2)
print(f'Decryption attempt with shift value 2: "{ciphertext2}" -> "{try2}"')

try3 = caesar.decrypt(ciphertext2, 3)
print(f'Decryption attempt with shift value 3: "{ciphertext2}" -> "{try3}"')

try4 = caesar.decrypt(ciphertext2, 4)
print(f'Decryption attempt with shift value 4: "{ciphertext2}" -> "{try4}"')
print(f'...')

try5 = caesar.decrypt(ciphertext2, 5)
print(f'Decryption attempt with shift value 5: "{ciphertext2}" -> "{try5}"\n\n')
````

## Railfence Cipher ##

Encrypting and decryption with the Railfence Cipher is a little more complicated because the rails form a zig-zag pattern that make it hard to do quick pattern matching.

### Railfence Encryption ###

Here's an example encrypting "For God so loved the world" with Railfence key lengths of 2, 3, and 4.

**Key=2**
```
For God so loved the world

F r G d s _ o e _ h _ o l    --> FrGds_oe_h_ol
 o _ o _ o l v d t e w r d   --> o_o_olvdtewrd

FrGds_oe_h_ol   o_o_olvdtewrd

FrGds_oe_h_olo_o_olvdtewrd
```

**Key=3**
```
For God so loved the world

F   G   s   o   _   _   l    --> FGso__l
 o _ o _ o l v d t e w r d   --> o_o_olvdtewrd
  r   d   _   e   h   o      --> rd_eho

FGso__l   o_o_olvdtewrd   rd_eho

FGso__lo_o_olvdtewrdrd_eho
```

**Key=4**
```
For God so loved the world

F     d     o     h     l    --> Fdohl
 o   o _   l v   t e   r d   --> oo_lvterd
  r G   s _   e _   _ o      --> rGs_e__o
   _     o     d     w       --> _odw

Fdohl   oo_lvterd   rGs_e__o   _odw

Fdohloo_lvterdrGs_e__o_odw
```

````python
rail_count_key = 4
plaintext3 = 'For God so loved the world'

rail_list3 = railfence.encrypt_show_rails(plaintext3, rail_count_key)

for rail in rail_list3:
    print(rail)
print('\n')
````

````python
ciphertext3 = railfence.encrypt(plaintext3, rail_count_key)

print(f'Encryption: "{plaintext3}" -> "{ciphertext3}"\n\n')
````

### Railfence Decryption ###

Decryption is the most complicated part of the Railfence cipher because the blocks of ciphertext is a sequence of rails--and the rails are all different lengths.

Here's an example decrypting the message "that he gave his one and only son", which has been encrypted to `te_n_shh_ehoedo_oa_gvi__nnyntasal` using four rails.

The ciphertext is 33 letters long, so four rails would come from a fence that looks like this:

```
*     *     *     *     *     *    -->  6
 *   * *   * *   * *   * *   * *   --> 11
  * *   * *   * *   * *   * *   *  --> 11
   *     *     *     *     *       -->  5
```

The means the ciphertext came from rails with length 6, 11, 11, and 5. These are the four rails:

```
te_n_shh_ehoedo_oa_gvi__nnyntasal -->

  1) te_n_s  
  2) hh_ehoedo_o
  3) a_gvi__nnyn
  4) tasal
```

Putting these rails into the proper rail pattern produces this fence:

```
t     e     _     n     _     s
 h   h _   e h   o e   d o   _ o
  a _   g v   i _   _ n   n y   n
   t     a     s     a     l
```

Reading the message along the zig-zag pattern produces the plaintext:

```
that he gave his one and only son
```

````python
cihpertext4 = 'te n shh ehoedo oa gvi  nnyntasal'

plaintext4 = railfence.decrypt(cihpertext4, rail_count_key)

print(f'Decryption: "{cihpertext4}" -> "{plaintext4}"\n\n')
````

### Railfence Cryptanalysis ###

The best way to retrieve the plaintext is to try decrypting with different length keys. If the key length is wrong, it will produce gibberish. But if you guess the key length correctly, it will generate a meaningful plaintext.

#### Attempt with Key=2 ####

```
te_n_shh_ehoedo_o   a_gvi__nnyntasal

t e _ n _ s h h ...
 a _ _ g v i _  ...

tae___ng_vsih_h ... (Gibberish)
```

```
  *   *   *   *   *   *   *   *    -->  8

te_n_shh_   ehoedo_oa_gvi__n   nyntasal

t   e   _   n   _
 e h o e d o _ o a
  n   y   n   t   a

tenheoye_dnon_to_aa ... (Gibberish again)
```

#### Attempt with Key=4 ####
```

  * *   * *   * *   * *   * *   *  --> 11

te_n_s   hh_ehoedo_o   a_gvi__nnyn   tasal
```

````python
ciphertext5 = 'te n shh ehoedo oa gvi  nnyntasal'

try2_rails = railfence.decrypt_show_rails(ciphertext5, 2)
print(f'Decryption attempt with shift value 2 "{ciphertext5}" ->')
for rail in try2_rails:
  print(f'  {rail}')

try3_rails = railfence.decrypt_show_rails(ciphertext5, 3)
print(f'\nDecryption attempt with shift value 3: "{ciphertext5}" -> "{try3}"')
for rail in try3_rails:
  print(f'  {rail}')

try4_rails = railfence.decrypt_show_rails(ciphertext5, 4)
print(f'\nDecryption attempt with shift value 4: "{ciphertext5}" -> "{try4}"')
for rail in try4_rails:
  print(f'  {rail}')
print('\n')
````
